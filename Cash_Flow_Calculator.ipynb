```python
import numpy_financial as npf

def solve_cash_flow(rate, cash_flows, pmt=0, periods=0, find_type="NPV"):
    # Convert percentage to decimal (e.g., 8 becomes 0.08)
    r = rate / 100

    if find_type == "NPV":
        # Net Present Value: Sum of all future money brought to today's value
        result = npf.npv(r, cash_flows)
        return f"The Net Present Value (NPV) is: ${result:,.2f}"

    elif find_type == "IRR":
        # Internal Rate of Return: The 'break-even' interest rate
        result = npf.irr(cash_flows)
        return f"The Internal Rate of Return (IRR) is: {result*100:.2f}%"

    elif find_type == "FV":
        # Future Value: What a series of payments will be worth later
        result = npf.fv(r, periods, -pmt, -cash_flows[0])
        return f"The Future Value (FV) is: ${result:,.2f}"

# --- EDIT YOUR DATA BELOW THIS LINE ---
interest_rate = 8          # The % rate
my_cash_flows = [-1000, 200, 300, 400, 500]  # Initial cost is negative, income is positive
# ---------------------------------------

print(solve_cash_flow(interest_rate, my_cash_flows, find_type="NPV"))
```
